```python
# ---GLOBAL DATA STORAGE---
# WE will store data in a list
# Example : [{'item' : 'Burger' , 'cost' : 100, 'category' : 'Food'}]
#expenses list data type
expenses = []

# ----FUNCTION 1: ADDING DATA -----
def add_expense(item_name,  amount,  category) :
    entry = {
        "item" : item_name,
        "cost" : amount,
        "category" : category
    }
    expenses.append(entry)
    print(f"Success! Added {item_name} to the list")

# -----FUNCTION 2 : ANALYSIS (TOTAL)
def get_total_expenses():
    total = 0
    #Loop through out list of dictionaries
    for expense in expenses :
        total += expense["cost"]
    return total

# FUNCTION 3 : ANALYSIS (   by category) ----
def get_category_total(target_category) :
    cat_total = 0
    for expense in expense :
        # Logic gate : AND condition (implicity)
        if expense['category'] == target_category :
            cat_total += expense['cost']
        return cat_total
    
    
```
