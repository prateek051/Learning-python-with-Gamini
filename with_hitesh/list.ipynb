```python
# lists are mutable -> which means we can change the list
# list are know as array in other languages

# list created using []
ingredients = ["water", "milk", "black tea"]

# adding new item to the list
ingredients.append("sugar")
print(f"Ingredients are : {ingredients}")

# removing the items fromt the list
ingredients.remove("water")
print(f"Ingredients are : {ingredients}")

spice_options = ["ginger", "cardamom"]
chai_ingredients = ["water", "milk"]

#Combining two list
chai_ingredients.extend(spice_options)
print(f"chai : {chai_ingredients}") # spice_option items are added at the starting  of the chai_ingredients list

# insert method in list
chai_ingredients.insert(2, "black tea") # 2 -> is the index number where we want to add the item in chai_ingreadients

print(chai_ingredients) # output -> ['water', 'milk', 'black tea', 'ginger', 'cardamom']

# pop method in list
last_added = chai_ingredients.pop() # pop mthod removes the last item from the list and hold that value 
# the pop value is stored in last_added variable
print(last_added) # output ->cardamom

# revers method in list
chai_ingredients.reverse()
print(chai_ingredients) # output -> ['ginger', 'black tea', 'milk', 'water']

# sort method in list 
chai_ingredients.sort() # the list is sorted in alpabetic order
print(chai_ingredients)  # output -> ['black tea', 'ginger', 'milk', 'water']

sugar_levels = [1, 2, 3, 4, 5, 10]
print(f"Maximum sugar level is : {max(sugar_levels)}") # max -> 10
print(f"Minimum sugar level is : {min(sugar_levels)}") # min -> 1

# Operator overloading -> meaning the operators not only used in their regulara operations but in adding list or multiplying the list with number like etc the this is know as operator overloading
base_liquid = ["water", "milk"]
extra_flaver = ["ginger"]

full_liquid_mix = base_liquid + extra_flaver
print(f"Liquid mix : {full_liquid_mix}")

strong_brew = ["black tea" , "ginger"] * 4
print(strong_brew)

raw_spice_data = bytearray(b"CINNAMON")
print(f"Bytes : {raw_spice_data}")

raw_spice_data = raw_spice_data.replace(b"CINNA", b"CARD")
print(f"Bytes : {raw_spice_data}")
```
