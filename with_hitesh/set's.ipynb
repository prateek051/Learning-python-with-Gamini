```python
# sets in python

essential_spices = {"cardamom", "ginger", "cinnamon"}
optional_spices = {"cloves", "ginger", "black pepeer"}

all_spices = essential_spices | optional_spices # "|" -> pipe operator also know as union operator
print(all_spices) # only unique items not duplicates of the items

common_spices = essential_spices & optional_spices
print(common_spices) # common items in the sets

only_in_essential = essential_spices - optional_spices
print(only_in_essential)

# member check or item check
print(f"Is 'cloves' in optional spices? {'cloves' in optional_spices}")

```
