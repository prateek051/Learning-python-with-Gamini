```python
# Tuples  are used with "()"
masala_spices = ("cardamom", "cloves", "cinnamon")

# Tuples are immutable - means they cannot be changed

(spice1, spice2, spice3) = masala_spices

# this is know as tuple unpacking
print(f"Main masala spices ; {spice1}, {spice2}, {spice3}")
```

```python
ginger_ratio, cadramom_ration = 2, 1
print(f"Ratio of G : {ginger_ratio} and C : {cadramom_ration}")

# swaping two values without third variable
ginger_ratio, cadramom_ration = cadramom_ration, ginger_ratio   # values are swaped
print(f"Ratio of G : {ginger_ratio} and C : {cadramom_ration}") 

# Membership
print(f"Is ginger in masala spices ? {"ginger" in masala_spices}") # membership test
print(f"Is ginger in masala spices ? {"cinnamon" in masala_spices}") # membership test
```
