```python
# Dictionaries in python
chai_order= dict (
    type = "Masala Chai",
    size = "Large",
    sugar = 2
)


print(chai_order)

chai_recipe = {}

#Adding the data to the dectionary
chai_recipe["base"] = "Black tea"
chai_recipe["liquid"] = "milk"

print(f"Recipe base : {chai_recipe['base']}")

print(chai_recipe)
# deleting the item from the dectionary
del chai_recipe["liquid"]
print(chai_recipe)



```

```python
chai_order = {"type" : "Ginger Chai", "size" : "Medium", "sugar": 1}

print(f"Order details (keys) : {chai_order.keys()}")
print(f"Order details (keys) : {chai_order.values()}")
print(f"Order details (keys) : {chai_order.items()}")

# Removing the whole item from the dectionaries
last_item = chai_order.popitem()
print(last_item)

print(chai_order)

extra_spices = {"cardamom" : "crushed", "ginger" : "sliced"}

chai_recipe.update(extra_spices)
print(chai_recipe)

print(f"Updated chai recipe: {chai_recipe}")
customer_note = chai_order.get("note", "NO Note")
print(f"customer_note is : {customer_note}")
```
